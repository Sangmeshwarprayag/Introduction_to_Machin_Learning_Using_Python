```python
import pandas as pd
df = pd.read_csv("iris.csv")
```

```python
print(df[df["species"].isin(["Iris-setosa", "setosa"])])
```

```python
print(df[(df["sepal_width"] > 3.5) &
         (df["species"].isin(["Iris-setosa", "setosa"]))])
```

```python
df.rename(columns={"sepal_width": "Sepal_Width"},
          inplace=True)
print(df.head())
```

```python
df["SepalArea"] = df["sepal_length"] * df["Sepal_Width"]
print(df.columns)
```

```python
df.drop_duplicates(inplace=True)
print(df)
```

```python
df["species"] = df["species"].replace(
    {"Iris-setosa": "Setosa", "setosa": "Setosa"}
)
print(df["species"].unique())
```

```python
df.to_csv("cleaned_iris.csv", index=False)
```
